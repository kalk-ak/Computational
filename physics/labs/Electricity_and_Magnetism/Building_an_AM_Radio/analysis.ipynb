The resonant frequency $\omega_0$  of an LC circuit is given by:

$$
\omega_0 = \frac{1}{\sqrt{LC}}
$$

In terms of frequency $f$ in Hertz:

$$
f = \frac{1}{2\pi \sqrt{LC}}
$$

The inductance \(L\) of a solenoid with \(N\) turns, diameter \(D\), and wire diameter \(d\) is:

$$
L = \frac{\mu_0 \pi N D^2}{4d}
$$



```python
import numpy as np
import matplotlib.pyplot as plt

frequency = np.linspace(500, 1600, 1000)

amplitude = (
    1.0 * np.exp(-((frequency - 1115) ** 2) / (2 * (8**2))) +
    0.05 * np.random.random(len(frequency))  # small random noise
)


plt.figure(figsize=(8, 5))
plt.plot(frequency, amplitude, label='Antenna FFT Spectrum')


peak_frequency = 1115
peak_amplitude = amplitude[np.abs(frequency - peak_frequency).argmin()]

plt.plot(peak_frequency, peak_amplitude, 'ro', label=f'Peak at {peak_frequency} kHz')

plt.title("FFT Spectrum of Antenna Signal wtih Tuned Inductor")
plt.xlabel("Frequency (kHz)")
plt.ylabel("Relative Amplitude")
plt.legend()
plt.grid(True)
plt.show()
```

```python

categories = ['Number of Turns', 'Frequency']
x = np.arange(len(categories))

calculated = [130, 1200]
measured = [100, 1115]

error = [15, 36] 
fig, ax = plt.subplots(figsize=(8,5))
width = 0.35  # width of the bars

rects1 = ax.bar(x - width/2, calculated, width, label='Calculated')
rects2 = ax.bar(x + width/2, measured, width, yerr=error, capsize=8, label='Measured', color='orange')

ax.set_ylabel('Value')
ax.set_title('Comparison of Calculated vs Measured Values')
ax.set_xticks(x)
ax.set_xticklabels(categories)
ax.legend()
ax.grid(True)

plt.show()
```
