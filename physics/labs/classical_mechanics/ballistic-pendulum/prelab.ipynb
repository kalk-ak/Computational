```python
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
```

# 1. Expression for the uncertainity

```python
r_cm, M, g, T = sp.symbols('r_cm M g T')
```

```python
I = (r_cm * M * g * T**2) / (4 * sp.pi**2)
I
```

```python
dr_cm, dM, dg, dT = sp.symbols('dr_cm dM dg dT')  # uncertainties
```

```python
uncertainty = sp.sqrt(
    (dr_cm / r_cm)**2 +
    (dM / M)**2 +
    (dg / g)**2 +
    (2 * dT / T)**2
)
uncertainty
```

```python

inertia_expression = I + uncertainty
sp.simplify(inertia_expression)


```

```python
inertia_expression = I - uncertainty
sp.simplify(inertia_expression)
```


To calculate the uncertainties in we first take the derivative and then multiply it with the associated uncertainity

### Vertical Axis Error Bars (Uncertainty in \( v^2 \))

 is calculated as:

$$
\delta (v^2) = 2v \cdot \delta v
$$

### Horizontal Axis Error Bars (Uncertainty in \( x^2 \))
 is calculated as:

$$
\delta (x^2) = 2x \cdot \delta x
$$
