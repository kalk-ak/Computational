## Exercise 2

**Problem 2.6** from Taylor's _"An Introduction to Error Analysis"_ (page 36)

"Each of two research groups discovers a new elementary particle.   The two reported masses are [reported below].   

"Draw an illustration like that in Figure 2.1 to represent these two measurements.   The question arises whether these two measurements could actually be of the same particle.   Based on the reported masses, would you say they are likely to be the same particle?  In particular, what is the discrepancy in the two measurements (assuming they really are measurements of the same mass)?"

```python
# Reported Mass 1
mass1  = 7.8*10**-27
dmass1 = 0.1*10**-27

# Reported Mass 2
mass2  = 7.0*10**-27
dmass2 = 0.2*10**-27

# Measurement units
mass_units = "kg"
```

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots()
ax.errorbar(1.2, mass1, yerr=dmass1, label='Mass of 1st Element',fmt='o',color='blue', capsize=7)
ax.errorbar(1.2, mass2, yerr=dmass2, label='Mass of 2nd Element',fmt='o', color='red', capsize=7)
ax.set_xlim(1.15, 1.5)
ax.set_xticks([])
ax.spines['top'].set_visible(False)
ax.spines['bottom'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.set_title('Mesurment of two masses and thier difference')
ax.set_ylabel(f"Mass in {mass_units} -->", size=12, color='black')
ax.annotate('', xy=(1.22, mass1), xytext=(1.22, mass2),
            arrowprops=dict(arrowstyle='<->', color='black', lw=1))
ax.text(1.222, (mass1 + mass2)/2, f'discrepancy = {abs(mass1 - mass2):.2g}', fontsize=10)
ax.legend()
```

I wouldn't say that the mesurments are of the same particle since there is no overlap between the measured data points and their uncertainities don't overlap


The desscrepancy between the two measurments is mass1 - mass2 which equals around 8 * 10^-28
